```python
import pandas as pd
import numpy as np
import xgboost as xgb
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error

# --- 1. THE GENERATOR ---
def generate_5g_data():
    """Generates 30 days of synthetic 5G traffic & energy data."""
    dates = pd.date_range(start='2024-01-01', periods=30*24, freq='h')
    
    # Simulate Traffic (Sine wave + Noise)
    hours = dates.hour
    load = (np.sin((hours - 6) * np.pi / 12) + 1) / 2 # Normalize 0-1
    load += np.random.normal(0, 0.1, len(dates))
    load = np.clip(load, 0.05, 1.0)
    
    # Simulate Energy (300W idle + usage)
    energy = 300 + (1200 * load) + np.random.normal(0, 10, len(dates))
    
    return pd.DataFrame({'Time': dates, 'Load': load, 'Energy': energy})

# --- 2. EXECUTE ---
print("📡 Generating Data...")
df = generate_5g_data()

# Feature Engineering
df['Hour'] = df['Time'].dt.hour
df['Lag_1h'] = df['Load'].shift(1)
df = df.dropna()

# --- 3. TRAIN MODEL ---
print("🧠 Training Model...")
X = df[['Hour', 'Lag_1h']]
y = df['Load']

model = xgb.XGBRegressor(n_estimators=100)
model.fit(X, y)

print(f"✅ SUCCESS! Model trained on {len(df)} rows.")
print(f"Sample Prediction (Next Hour Load): {model.predict(X.iloc[0:1])[0]:.2f}")
```
