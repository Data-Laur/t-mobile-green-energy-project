```python
# 1. Install necessary libraries (this fixes the "ModuleNotFoundError")
!pip install pandas numpy matplotlib seaborn xgboost scikit-learn

import os
import shutil
import pandas as pd
import numpy as np

# --- CONFIGURATION ---
# We map file extensions to their new homes
FOLDERS = {
    "data/raw": [".csv"],
    "data/processed": [],
    "notebooks": [".ipynb"],
    "images": [".png", ".jpg", ".jpeg"],
    "src": [".py"]
}
# Files to ignore (so we don't move the script itself or git files)
IGNORE_FILES = ["requirements.txt", ".gitignore", "README.md", "LICENSE"]

# 2. Create Directories
print("👷 Creating Project Structure...")
for folder in FOLDERS:
    os.makedirs(folder, exist_ok=True)

# 3. Move Files (Smart Search)
print("🧹 Moving Files...")
current_notebook_name = "00_Repo_Cleanup.ipynb" # We won't move this one yet

for filename in os.listdir("."):
    if filename in IGNORE_FILES or filename == current_notebook_name or os.path.isdir(filename):
        continue
        
    # Determine file type and destination
    ext = os.path.splitext(filename)[1].lower()
    for folder, extensions in FOLDERS.items():
        if ext in extensions:
            # Check if it's a "Project 2" file we want to delete
            if filename in ["tools.py", "agent.py", "main.py", "load_test.py", "cleanup_repo.py"]:
                print(f"   ❌ Deleting temp file: {filename}")
                os.remove(filename)
                break
            
            # Move the file
            try:
                shutil.move(filename, os.path.join(folder, filename))
                print(f"   Moved {filename} -> {folder}/")
            except Exception as e:
                # If file already exists in destination, skip it
                print(f"   Skipped {filename} (Already moved)")
            break

# 4. Generate the "Missing" Digital Twin Dataset
print("🧬 Generating 'Digital Twin' Synthetic Dataset...")
np.random.seed(42)
n_samples = 10000
df_synthetic = pd.DataFrame({
    # FIX: Changed 'H' to 'h' for compatibility with new Pandas versions
    'Time': pd.date_range(start='2024-01-01', periods=n_samples, freq='h'),
    'Traffic_Load': np.random.beta(2, 5, n_samples),  # Skewed distribution
    'Antennas': np.random.choice([4, 32, 64], n_samples, p=[0.8, 0.15, 0.05])
})
# The Physics Equation (Your "Digital Twin" Logic)
df_synthetic['Predicted_Energy'] = (
    15 +  # Base Load
    (df_synthetic['Traffic_Load'] * 20) + # Traffic Load
    (df_synthetic['Antennas'] * 0.5)      # Hardware Factor
)
# Save it
df_synthetic.to_csv("data/processed/synthetic_digital_twin.csv", index=False)
print("   ✅ Created data/processed/synthetic_digital_twin.csv")

# 5. Create requirements.txt
reqs = """pandas
numpy
matplotlib
seaborn
xgboost
scikit-learn
jupyter
"""
with open("requirements.txt", "w") as f:
    f.write(reqs)
print("   ✅ Created requirements.txt")

print("\n✨ Clean up complete! Check your file explorer on the left.")
```
